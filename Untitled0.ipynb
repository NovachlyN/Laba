В качестве набора данных мы будем использовать набор данных для классификации мобильных телефонов по ценовым категориям, доступный на платформе Kaggle: Mobile Price Classification - https://www.kaggle.com/datasets/iabhishekofficial/mobile-price-classification?resource=download

Эта задача является актуальной для производителей мобильных устройств и потребителей, позволяя на основе технических характеристик предсказывать ценовой сегмент телефона (бюджетный, средний, флагман и т.д.). Такая информация может быть полезна при разработке маркетинговых стратегий, позиционировании продуктов и анализе рынка.

Датасет представлен в одном файле:

train.csv – обучающая выборка, содержащая 15 записей с техническими характеристиками телефонов и целевой переменной.

Каждая запись содержит следующие колонки:
- **battery_power** – ёмкость аккумулятора (мА·ч)
- **blue** – поддержка Bluetooth (0/1)
- **clock_speed** – тактовая частота процессора (ГГц)
- **dual_sim** – поддержка двух SIM (0/1)
- **fc** – разрешение фронтальной камеры (Мп)
- **four_g** – поддержка 4G (0/1)
- **int_memory** – внутренняя память (ГБ)
- **m_dep** – толщина корпуса (см)
- **mobile_wt** – вес телефона (г)
- **n_cores** – количество ядер процессора
- **pc** – разрешение основной камеры (Мп)
- **px_height** – высота дисплея в пикселях
- **px_width** – ширина дисплея в пикселях
- **ram** – оперативная память (МБ)
- **sc_h** – высота экрана (см)
- **sc_w** – ширина экрана (см)
- **talk_time** – время разговора (часы)
- **three_g** – поддержка 3G (0/1)
- **touch_screen** – сенсорный экран (0/1)
- **wifi** – поддержка Wi-Fi (0/1)
- **price_range** – целевая переменная,ценовая категория (0 – бюджетный, 1 – средний, 2 – выше среднего, 3 – флагман.)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Загрузка
df = pd.read_csv('train.csv')

# Если есть нечисловые столбцы, оставляем только числа
numeric_cols = df.select_dtypes(include=['number']).columns
df_numeric = df[numeric_cols]

# Удаляем строки с пропусками (если есть)
df_numeric = df_numeric.dropna()

# Строим pairplot, можно добавить hue='price_range' для раскраски
sns.pairplot(df_numeric, hue='price_range', palette='deep')
plt.show()
```
